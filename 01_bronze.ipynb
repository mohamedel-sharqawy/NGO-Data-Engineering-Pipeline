```python

from pyspark.sql import functions as F

cases_raw = (
    spark.read.option("header", True)
    .option("inferSchema", False)
    .csv(f"{raw_path}cases.csv")
    .withColumn("_ingested_at", F.current_timestamp())
    .withColumn("_source_file", F.col("_metadata.file_path"))
)

cases_raw.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable(bronze_cases)

services_raw = (
    spark.read.option("header", True)
    .option("inferSchema", False)
    .csv(f"{raw_path}services.csv")
    .withColumn("_ingested_at", F.current_timestamp())
    .withColumn("_source_file", F.col("_metadata.file_path"))
)

services_raw.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable(bronze_services)

display(spark.table(bronze_cases).limit(5))
```
